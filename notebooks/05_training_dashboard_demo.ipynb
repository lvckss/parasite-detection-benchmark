```python
from datetime import datetime
from IPython.display import display
from parasite_detection.paths import RESULTS
from parasite_detection.dashboard import TrainingDashboard
from parasite_detection.dashboard.demo import simulate_training

# usamos una carpeta nueva para no pisar una demo anterior
run_name = "demo-" + datetime.now().strftime("%Y%m%d-%H%M%S-%f")
report_dir = RESULTS / "reports" / run_name
epochs = 12
```

```python
# las flechas cambian de gráfica mientras llegan las métricas
history = simulate_training(dashboard, report_dir, epochs=epochs, delay=0.06)
```
